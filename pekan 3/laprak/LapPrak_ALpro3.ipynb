## 2.1 Dasar Teori
### 2.1.1 Python
Python merupakan salah satu bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana dan mudah dipahami. Python banyak digunakan dalam berbagai bidang, seperti pengembangan perangkat lunak, pengolahan data, dan pembelajaran pemrograman. Karena sintaksnya relatif sederhana, Python juga sesuai digunakan untuk mempelajari konsep dasar pemrograman.

Dalam Python terdapat berbagai sintaks dasar, seperti fungsi print() yang digunakan untuk menampilkan informasi ke layar dan komentar yang digunakan untuk memberikan keterangan pada kode program. Python juga menggunakan string untuk menyimpan atau menampilkan data berupa teks. Beberapa string dapat digabungkan menggunakan operator +. Selain itu, Python memiliki escape character, seperti \\, yang dapat digunakan untuk menghasilkan karakter atau format tertentu dalam output.

### 2.1.2 Operasi Aritmatika
Operator aritmatika digunakan untuk melakukan operasi matematika terhadap nilai numerik. Python menyediakan beberapa operator aritmatika, yaitu +, -, *, /, //, %, dan **.

1. Penjumlahan (+) digunakan untuk menjumlahkan dua nilai.
2. Pengurangan (-) digunakan untuk mencari selisih antara dua nilai.
3. Perkalian (*) digunakan untuk mengalikan dua nilai.
4. Pembagian (/) digunakan untuk membagi suatu nilai dan menghasilkan hasil pembagian dalam bentuk floating-point.
5. Pembagian bulat (//) digunakan untuk melakukan floor division atau pembagian dengan mengambil nilai floor dari hasil pembagian.
6. Modulus (%) digunakan untuk memperoleh sisa hasil pembagian.
7. Perpangkatan (**) digunakan untuk menghitung suatu nilai yang dipangkatkan dengan nilai lainnya.

Operator aritmatika banyak digunakan dalam pengolahan data numerik, seperti perhitungan nilai, luas, volume, dan berbagai proses matematika lainnya.

### 2.1.3 Operasi Perbandingan
Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil dari operasi ini berupa nilai Boolean, yaitu True atau False. Operator perbandingan dapat digunakan untuk mengetahui hubungan antara dua nilai dan sering digunakan dalam kondisi atau pengambilan keputusan pada program.

Operator perbandingan dalam Python meliputi:
1. Lebih besar dari (>) digunakan untuk memeriksa apakah nilai pertama lebih besar daripada nilai kedua.
2. Lebih kecil dari (<) digunakan untuk memeriksa apakah nilai pertama lebih kecil daripada nilai kedua.
3. Lebih besar atau sama dengan (>=) digunakan untuk memeriksa apakah nilai pertama lebih besar atau sama dengan nilai kedua.
4. Lebih kecil atau sama dengan (<=) digunakan untuk memeriksa apakah nilai pertama lebih kecil atau sama dengan nilai kedua.
5. Sama dengan (==) digunakan untuk memeriksa apakah dua nilai memiliki nilai yang sama.
6. Tidak sama dengan (!=) digunakan untuk memeriksa apakah dua nilai berbeda.

Python juga mendukung perbandingan berantai, misalnya 0 < x < 100, yang dapat digunakan untuk memeriksa apakah suatu nilai berada dalam rentang tertentu.

### 2.1.4 Operasi Logika
Operator logika digunakan untuk menggabungkan atau memodifikasi kondisi yang menghasilkan nilai Boolean. Operator logika utama dalam Python adalah and, or, dan not.
1. and digunakan untuk menghubungkan dua kondisi dan menghasilkan True apabila kedua kondisi bernilai benar.
2. or menghasilkan True apabila salah satu atau kedua kondisi bernilai benar.
3. not digunakan untuk membalik nilai kebenaran suatu kondisi. Jika nilai awal True, maka not menghasilkan False, begitu juga sebaliknya.

Dalam praktikum, operasi logika juga dapat digunakan untuk menerapkan XOR (Exclusive OR) menggunakan perbandingan != pada dua nilai Boolean. XOR menghasilkan True ketika kedua nilai berbeda.
Operator logika sangat penting dalam pemrograman karena dapat digunakan untuk menggabungkan beberapa kondisi sebelum program menentukan suatu tindakan.

### 2.1.5 Operasi Assignment
Operator assignment digunakan untuk memberikan nilai kepada variabel atau memperbarui nilai yang telah tersimpan dalam variabel. Operator assignment dasar dalam Python adalah =. Selain itu, terdapat operator assignment gabungan yang menggabungkan assignment dengan operasi tertentu.

Beberapa operator assignment yang digunakan dalam praktikum adalah:
1. = digunakan untuk memberikan nilai kepada variabel.
2. += digunakan untuk menambahkan nilai ke variabel.
3. -= digunakan untuk mengurangi nilai variabel.
4. *= digunakan untuk mengalikan nilai variabel.
5. /= digunakan untuk membagi nilai variabel.
6. //= digunakan untuk melakukan pembagian bulat dan menyimpan hasilnya.
7. %= digunakan untuk memperoleh sisa bagi dan menyimpan hasilnya.
8. **= digunakan untuk melakukan perpangkatan dan menyimpan hasilnya.

Penggunaan assignment gabungan membuat penulisan operasi menjadi lebih ringkas. Misalnya, x += y memiliki fungsi yang setara dengan memberikan hasil x + y kembali ke variabel x.

### 2.1.6 Operasi Lainnya
#### 2.1.6.1 Operator Keanggotaan
Operator keanggotaan atau membership operator digunakan untuk memeriksa apakah suatu nilai terdapat dalam suatu objek atau kumpulan data. Python menyediakan operator in dan not in.
Operator in menghasilkan True apabila nilai yang dicari terdapat dalam objek. Sebaliknya, not in menghasilkan True apabila nilai tersebut tidak terdapat dalam objek.
Contohnya, jika terdapat sebuah list yang berisi beberapa angka, operator in dapat digunakan untuk memeriksa apakah angka tertentu terdapat dalam list tersebut. Operasi ini berguna dalam proses pencarian dan pemeriksaan data.

#### 2.1.6.2 Operator Identitas
Operator identitas atau identity operator digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama. Python menggunakan operator is dan is not untuk melakukan pemeriksaan identitas.
1. is menghasilkan True apabila dua variabel merujuk pada objek yang sama.
2. is not menghasilkan True apabila dua variabel tidak merujuk pada objek yang sama.
Operator identitas berbeda dengan operator ==. Operator == digunakan untuk membandingkan nilai, sedangkan is digunakan untuk membandingkan identitas objek. Dengan demikian, dua objek dapat memiliki isi atau nilai yang sama tetapi belum tentu merupakan objek yang sama.

### 2.1.7 Operasi Bitwise
Operator bitwise digunakan untuk melakukan operasi terhadap bilangan berdasarkan representasi binernya. Pada operasi ini, bilangan diproses berdasarkan bit-bit penyusunnya. Python menyediakan operator bitwise seperti &, |, ^, ~, <<, dan >>.

Jenis operasi bitwise meliputi:
1. Bitwise AND (&) menghasilkan bit 1 apabila kedua bit yang dibandingkan bernilai 1.
2. Bitwise OR (|) menghasilkan bit 1 apabila salah satu atau kedua bit bernilai 1.
3. Bitwise XOR (^) menghasilkan bit 1 apabila kedua bit yang dibandingkan memiliki nilai berbeda.
4. Bitwise NOT (~) digunakan untuk membalik bit suatu bilangan.
5. Left Shift (<<) digunakan untuk menggeser bit ke arah kiri berdasarkan jumlah pergeseran yang ditentukan.
6. Right Shift (>>) digunakan untuk menggeser bit ke arah kanan berdasarkan jumlah pergeseran yang ditentukan.

Dalam praktikum, fungsi bin() digunakan untuk menampilkan bilangan dalam bentuk biner. Hal tersebut membantu melihat representasi biner dari nilai sebelum dan sesudah dilakukan operasi bitwise.

3. Variabel angka1_3007 dan angka2_3007 digunakan sebagai nilai awal yang akan diproses.
4. Variabel hasil_3007 digunakan untuk menyimpan nilai dan hasil dari setiap operasi assignment.
5. Operator = digunakan untuk memberikan atau menyalin nilai dari angka1_3007 ke hasil_3007.
6. Operator += digunakan untuk menambahkan nilai angka2_3007 ke hasil_3007.
7. Operator -= digunakan untuk mengurangi nilai hasil_3007 dengan angka2_3007.
8. Operator *= digunakan untuk mengalikan nilai hasil_3007 dengan angka2_3007.
9. Operator /= digunakan untuk membagi nilai hasil_3007 dengan angka2_3007.
10. Operator //= digunakan untuk melakukan pembagian bulat dan menyimpan hasilnya kembali ke hasil_3007.
11. Operator %= digunakan untuk mendapatkan sisa hasil bagi dan menyimpan hasilnya ke hasil_3007.

### 2.2.6 Program Bitwise_2611533007

```python
print("\n======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_3007 = int(input("Masukkan angka bitwise-1: "))
angka2_3007 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner:")
print("angka1_3007 =", angka1_3007, "| biner =", bin(angka1_3007))
print("angka2_3007 =", angka2_3007, "| biner =", bin(angka2_3007))

#Bitwise AND
hasil_3007 = angka1_3007 & angka2_3007
print("\nBitwise AND (&)")
print(angka1_3007, "&", angka2_3007, "=", hasil_3007)
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

#Bitwise OR
hasil_3007 = angka1_3007 | angka2_3007
print("\nBitwise OR (|)")
print(angka1_3007, "|", angka2_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

#Bitwise XOR
hasil_3007 = angka1_3007 ^ angka2_3007
print("\nBitwise XOR (^)")
print(angka1_3007, "^", angka2_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

#Bitwise NOT
hasil_3007 = ~angka1_3007
print("\nBitwise NOT (~)")
print("~", angka1_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

#Bitwise geser kiri
jumlah_geser_3007 = int(input("\nMasukkan jumlah pergeseran bit: "))
hasil_3007 = angka1_3007 << jumlah_geser_3007
print("\nBitwise Geser Kiri (<<)")
print(angka1_3007, "<<", jumlah_geser_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
print("Biner hasil (8 bit) =", format(hasil_3007, '08b'))

#Bitwise geser kanan
hasil_3007 = angka1_3007 >> jumlah_geser_3007
print("\nBitwise Geser Kanan (>>)")
print(angka1_3007, ">>", jumlah_geser_3007, "=", hasil_3007)
print("Biner hasil =", bin(hasil_3007))
```

**Penjelasan :**
1. input() digunakan untuk menerima dua angka dan jumlah pergeseran bit dari pengguna.
2. int() digunakan untuk mengubah input menjadi bilangan bulat.
3. Fungsi bin() digunakan untuk mengubah bilangan desimal menjadi bilangan biner.
4. Operator Bitwise AND (&) digunakan untuk membandingkan setiap bit dari dua bilangan dan menghasilkan 1 jika kedua bit bernilai 1.
5. Operator Bitwise OR (|) menghasilkan 1 jika salah satu atau kedua bit bernilai 1.
6. Operator Bitwise XOR (^) menghasilkan 1 jika kedua bit memiliki nilai yang berbeda.
7. Operator Bitwise NOT (~) digunakan untuk membalik setiap bit dari suatu bilangan.
8. Operator geser kiri (<<) digunakan untuk menggeser bit ke arah kiri sesuai jumlah pergeseran yang ditentukan.
9. Operator geser kanan (>>) digunakan untuk menggeser bit ke arah kanan sesuai jumlah pergeseran yang ditentukan.
10. Variabel jumlah_geser_3007 digunakan untuk menyimpan jumlah bit yang akan digeser.
11. Variabel hasil_3007 digunakan untuk menyimpan hasil dari setiap operasi bitwise.
12. Fungsi format(hasil_3007, '08b') digunakan untuk menampilkan hasil bilangan dalam bentuk biner dengan format 8 bit.
