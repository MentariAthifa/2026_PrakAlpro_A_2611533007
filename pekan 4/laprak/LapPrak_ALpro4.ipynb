## 2.1 Dasar Teori
### 2.1.1 Python
Python merupakan salah satu bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana dan mudah dipahami. Python banyak digunakan dalam berbagai bidang, seperti pengembangan perangkat lunak, pengolahan data, dan pembelajaran pemrograman. Karena sintaksnya relatif sederhana, Python juga sesuai digunakan untuk mempelajari konsep dasar pemrograman.

Dalam Python terdapat berbagai sintaks dasar, seperti fungsi print() yang digunakan untuk menampilkan informasi ke layar dan komentar yang digunakan untuk memberikan keterangan pada kode program. Python juga menggunakan string untuk menyimpan atau menampilkan data berupa teks. Beberapa string dapat digabungkan menggunakan operator +. Selain itu, Python memiliki escape character, seperti \\, yang dapat digunakan untuk menghasilkan karakter atau format tertentu dalam output.

### 2.1.2 Conditional
Conditional atau percabangan merupakan konsep dalam pemrograman yang digunakan untuk menentukan tindakan berdasarkan suatu kondisi. Kondisi akan dievaluasi dan menghasilkan nilai benar atau salah, kemudian program menentukan blok perintah yang akan dijalankan berdasarkan hasil tersebut. Penggunaan conditional membuat program mampu memberikan keluaran yang berbeda meskipun menggunakan struktur program yang sama, bergantung pada nilai atau kondisi yang diberikan.

Dalam Python, conditional dapat diterapkan menggunakan beberapa bentuk, yaitu if, if else, if elif else, dan match case. Pemilihan bentuk percabangan disesuaikan dengan jumlah dan jenis kondisi yang perlu diperiksa dalam program.

### 2.1.3 If
Pernyataan if merupakan bentuk dasar dari conditional pada Python. Pernyataan ini digunakan untuk menjalankan suatu blok kode apabila kondisi yang diperiksa bernilai benar (True). Jika kondisi bernilai salah (False), blok kode tersebut akan dilewati dan program melanjutkan eksekusi ke perintah berikutnya.

Bentuk umum if terdiri dari kata kunci if, kondisi yang akan diperiksa, tanda titik dua (:), dan blok kode yang ditulis dengan indentasi. Pernyataan if sesuai digunakan ketika program hanya perlu memeriksa satu kondisi tertentu.

### 2.1.4 If Else
Pernyataan if else digunakan ketika program memiliki dua kemungkinan tindakan berdasarkan suatu kondisi. Jika kondisi pada if bernilai benar, program menjalankan blok kode pada if. Sebaliknya, jika kondisi bernilai salah, program menjalankan blok kode pada else.

Penggunaan if else memungkinkan program memberikan alternatif ketika kondisi utama tidak terpenuhi. Dengan demikian, program dapat menangani dua keadaan yang saling berlawanan, seperti kondisi benar dan salah atau memenuhi dan tidak memenuhi suatu persyaratan.

### 2.1.5 If Elif Else
Pernyataan if elif else digunakan ketika terdapat beberapa kondisi yang perlu diperiksa secara berurutan. if digunakan untuk memeriksa kondisi pertama, sedangkan elif (else if) digunakan untuk memeriksa kondisi berikutnya. Sementara itu, else digunakan untuk menangani keadaan ketika tidak ada kondisi sebelumnya yang terpenuhi.

Python memungkinkan penggunaan lebih dari satu elif dalam satu rangkaian percabangan. Ketika suatu kondisi terpenuhi, blok kode yang sesuai akan dijalankan dan rangkaian percabangan tersebut selesai diproses. Struktur ini dapat digunakan, misalnya, untuk menentukan kategori berdasarkan rentang nilai atau menentukan pilihan berdasarkan beberapa kondisi.

### 2.1.6 Multi If
Multi if merupakan penggunaan beberapa pernyataan if secara terpisah dalam satu program. Setiap pernyataan if akan memeriksa kondisinya masing-masing secara independen. Oleh karena itu, apabila terdapat beberapa kondisi yang bernilai benar, lebih dari satu blok perintah dapat dijalankan.

Multi if berbeda dengan struktur if elif else. Pada if elif else, kondisi diperiksa sebagai satu rangkaian sehingga ketika suatu kondisi terpenuhi, kondisi berikutnya dalam rangkaian tersebut tidak dipilih. Sementara itu, pada beberapa if yang berdiri sendiri, setiap kondisi tetap diperiksa sehingga beberapa kondisi dapat terpenuhi dan menghasilkan beberapa keluaran.

### 2.1.7 Match Case
Match case merupakan struktur percabangan pada Python yang digunakan untuk melakukan pattern matching terhadap suatu nilai atau objek. Pernyataan match mengevaluasi nilai yang diberikan, kemudian mencocokkannya dengan pola yang terdapat pada setiap case. Jika suatu pola cocok, blok kode pada case tersebut akan dijalankan.

Pernyataan match diperkenalkan pada Python 3.10. Struktur ini dapat digunakan ketika program perlu membandingkan satu nilai dengan beberapa kemungkinan pola. Selain pencocokan nilai sederhana, match case juga mendukung pola yang lebih kompleks.

Dalam match case, case _ dapat digunakan sebagai pola umum (wildcard) untuk menangani nilai yang tidak sesuai dengan pola-pola sebelumnya. Selain itu, sebuah case dapat memiliki guard berupa kondisi if yang digunakan untuk menentukan apakah blok tersebut boleh dijalankan setelah pola berhasil dicocokkan.

### 2.1.8 Indentasi pada Conditional Python
Indentasi merupakan bagian penting dalam penulisan struktur conditional pada Python. Indentasi digunakan untuk menunjukkan blok kode yang termasuk dalam suatu pernyataan. Pada struktur if, elif, else, dan case, perintah yang berada di dalam blok harus memiliki indentasi yang sesuai.

Python menggunakan indentasi untuk menentukan batas suatu blok kode, sehingga penulisan indentasi yang tidak tepat dapat menyebabkan kesalahan pada program. Oleh karena itu, penggunaan indentasi yang konsisten diperlukan agar struktur percabangan dapat dikenali dan dijalankan dengan benar oleh Python.

### 2.1.9 Penerapan Conditional dalam Program
Conditional dapat diterapkan pada berbagai permasalahan pemrograman yang membutuhkan pengambilan keputusan. Contohnya adalah menentukan kelulusan berdasarkan nilai, menentukan kategori berdasarkan jumlah atau kondisi tertentu, memeriksa apakah suatu persyaratan telah terpenuhi, serta menentukan pilihan berdasarkan input pengguna.

Dengan menggunakan conditional, program dapat menjadi lebih dinamis karena keluaran tidak hanya bergantung pada perintah yang ditulis secara berurutan, tetapi juga dapat menyesuaikan diri terhadap kondisi yang terjadi. Pemahaman terhadap if, if else, if elif else, multi if, dan match case menjadi dasar untuk membangun logika program yang lebih terstruktur dan mampu menangani berbagai kemungkinan kondisi.

### 2.2.1 Program if1_2611533007

```python
ipk_3007 = float(input("Input IPK anda = "))

if ipk_3007 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK" + str(ipk_3007))

print("Program Selesai")
```

Fungsi str() digunakan untuk mengubah nilai IPK dari tipe float menjadi tipe string agar dapat digabungkan dengan teks menggunakan operator +.<br>
4. print("Program Selesai") digunakan untuk menampilkan pesan bahwa program telah selesai dijalankan. Perintah ini berada di luar blok if, sehingga akan tetap dijalankan meskipun kondisi ipk_3007 > 2.75 tidak terpenuhi.<br>

### 2.2.7 Program match_case_2611533007

```python
bulan_3007 = int(input("Masukkan angka bulan (1-12): "))

match bulan_3007:
    case 1 :
        print("Januari")
    case 2 :
        print("Februari")
    case 3 :
        print("Maret")
    case 4 :
        print("April")
    case 5 :
        print("Mei")
    case 6 :
        print("Juni")
    case 7 :
        print("Juli")
    case 8 :
        print("Agustus")
    case 9 : 
        print("September")
    case 10 :
        print("Oktober")
    case 11 :
        print("November")
    case 12 :
        print("Desember")
    case _ :
        print("Angka Tidak Valid")
```

**Penjelasan :**
1. bulan_3007 = int(input(...)) digunakan untuk menerima input angka bulan dalam bentuk bilangan bulat.<br>
2. match bulan_3007: digunakan untuk mencocokkan nilai input dengan beberapa pilihan case yang tersedia.<br>
3. case 1 sampai case 12 digunakan untuk menentukan nama bulan berdasarkan angka yang dimasukkan.<br>
4. Setiap case yang sesuai akan menjalankan perintah print() untuk menampilkan nama bulan.<br>
5. case _: digunakan sebagai kondisi umum apabila angka yang dimasukkan tidak sesuai dengan angka 1 sampai 12.<br>
6. print("Angka Tidak Valid") menampilkan pesan apabila input berada di luar rentang angka bulan yang tersedia.

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK === ")

nama_3007 = input("Masukkan Nama Pengunjung :")
umur_3007 = int(input("Masukkan Umur Anda : "))
sim_c_3007 =  input("Apakah Anda Sudah Punya SIM C? (y/t) :").strip().lower()

print("Pilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_3007 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_3007 = int(input("Masukkan jumlah tiket           : "))

is_member_3007 = input("Apakah Anda member? (y/t)       : ")
kode_promo_valid_3007 = input("Apakah kode promo valid? (y/t)  : ")

if jumlah_tiket_3007 < 0:
    print("Jumlah Tiket Tidak Valid")

harga_paket_3007 = 0

match paket_3007:
    case 1:
        nama_paket_3007 = "Wahana Safari Rimba"
        harga_paket_3007 = 50000
    case 2:
        nama_paket_3007 = "Wahana Arung Jeram"
        harga_paket_3007 = 75000
    case 3:
        nama_paket_3007 = "Wahana Motor ATV Ekstrim"
        harga_paket_3007 = 120000
    case 4:
        nama_paket_3007 = "Wahana Roller Coaster Kilat"
        harga_paket_3007 = 100000
    case 5:
        nama_paket_3007 = "Wahana All-Access VIP"
        harga_paket_3007 = 220000
    case _:
        print("Wahana Tidak Valid")

print()
print("=== KELAYAKAN PENGENDARA WAHANA ===")

if paket_3007 == 3 :
 if umur_3007 >= 17 and sim_c_3007 == 'y':
    print("Status Akses : Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
 elif umur_3007 >= 17 and sim_c_3007 != 'y':
    print("Status Akses : Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
 elif umur_3007 < 17 and sim_c_3007 == 'y':
    print("Status Akses : Identitas tidak valid: Belum cukup umur memiliki SIM.")
 else:
    print("Status Akses : Anda belum cukup umur dan tidak boleh bawa motor ATV.")
else:
    if umur_3007 >= 10 :
        print("Anda sudah cukup umur untuk menaiki wahana {nama_paket_3007}")
    else:
        print("Anda belum cukup umur untuk menaiki wahana {nama_paket_3007}")     


subtotal_3007 = harga_paket_3007 * jumlah_tiket_3007
total_diskon_persen_3007 = 0

if subtotal_3007 > 200000:
    total_diskon_persen_3007 += 10 #Diskon belanja besar

if is_member_3007 in ['y', 'ya']:
    total_diskon_persen_3007 += 5 #Diskon member

if kode_promo_valid_3007 in ['y', 'ya']:
    total_diskon_persen_3007 += 15 #Diskon voucher

if jumlah_tiket_3007 >= 5:
    total_diskon_persen_3007 += 5 #Diskon tambahan rombongan

nominal_diskon_3007 = subtotal_3007 * (total_diskon_persen_3007 / 100)
total_bayar_3007 = subtotal_3007 - nominal_diskon_3007

print("\n--- Rincian Pembayaran ---")
print(f"Subtotal Belanja : Rp {subtotal_3007:,.0f}")
print(f"Total Diskon : {total_diskon_persen_3007}% (Rp {nominal_diskon_3007:,.0f})")
print(f"Total Bayar : Rp {total_bayar_3007:,.0f}")

if total_bayar_3007 > 300000:
    print("Catatan Layanan : Selamat! Anda berhak mendapatkan souvenir gratis.")
else:
    print("Catatan Layanan : Terima kasih telah berkunjung.")

print("Program selesai")
```
