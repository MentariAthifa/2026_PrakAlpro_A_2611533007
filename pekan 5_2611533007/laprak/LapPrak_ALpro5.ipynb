## 2.1 Dasar Teori
### 2.1.1 Python
Python merupakan salah satu bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana dan mudah dipahami. Python banyak digunakan dalam berbagai bidang, seperti pengembangan perangkat lunak, pengolahan data, dan pembelajaran pemrograman. Karena sintaksnya relatif sederhana, Python juga sesuai digunakan untuk mempelajari konsep dasar pemrograman.

Dalam Python terdapat berbagai sintaks dasar, seperti fungsi print() yang digunakan untuk menampilkan informasi ke layar dan komentar yang digunakan untuk memberikan keterangan pada kode program. Selain itu, Python memiliki escape character, seperti \\, yang dapat digunakan untuk menghasilkan karakter atau format tertentu dalam output.

### 2.1.2 Perulangan dalam Python

Perulangan atau looping merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk menjalankan suatu perintah secara berulang selama kondisi atau batas tertentu terpenuhi. Perulangan membantu programmer mengurangi penulisan kode yang sama secara berulang sehingga program menjadi lebih sederhana dan efisien. Dalam Python, salah satu bentuk perulangan yang sering digunakan adalah for.

Perulangan for digunakan untuk melakukan iterasi terhadap suatu urutan data atau rentang nilai. Pada program yang dibuat, perulangan for digunakan untuk menghasilkan nilai dari 1 sampai batas yang dimasukkan oleh pengguna.

### 2.1.3 Perulangan for

Perulangan for merupakan struktur kontrol yang digunakan untuk mengulang blok kode berdasarkan elemen atau nilai yang terdapat dalam suatu sequence. Secara umum, bentuk perulangan for dalam Python adalah:

for variabel in range():

    perintah

Pada program, perulangan pertama dituliskan sebagai:

for line in range(1, batas_3007 + 1):

Perintah tersebut membuat variabel line menerima nilai secara berurutan mulai dari 1 hingga nilai batas_3007. Penggunaan + 1 dilakukan karena batas akhir pada fungsi range() tidak ikut dihitung.

### 2.1.4 Fungsi range

range() merupakan fungsi bawaan Python yang digunakan untuk menghasilkan deretan bilangan secara berurutan. Fungsi ini dapat digunakan untuk menentukan nilai awal, nilai akhir, dan interval perulangan.

Bentuk umum range() adalah:

range(start, stop, step)

Parameter start menunjukkan nilai awal, stop menunjukkan batas akhir yang tidak termasuk dalam hasil, sedangkan step menunjukkan perubahan nilai pada setiap iterasi.

Pada kode yang digunakan:
range(1, batas_3007 + 1)
perulangan dimulai dari angka 1 dan berhenti sebelum mencapai batas_3007 + 1. Dengan demikian, nilai terakhir yang diproses adalah batas_3007.

### 2.1.5 Perulangan Bersarang (Nested Loop)

Nested loop adalah perulangan yang berada di dalam perulangan lainnya. Perulangan ini digunakan ketika suatu proses membutuhkan pengulangan di dalam setiap iterasi dari perulangan utama.

Pada program terdapat dua buah perulangan for. Perulangan kedua berada di dalam perulangan pertama:

for line in range(1, batas_3007 + 1):

    for j in range(1, (-1 * line + batas_3007) + 1):
    
        print(".", end="")

Artinya, setiap kali perulangan line dijalankan, perulangan j juga akan dijalankan. Jumlah pengulangan j bergantung pada nilai line dan batas_3007.

Konsep nested loop banyak digunakan untuk membuat pola, mengolah data berbentuk baris dan kolom, serta melakukan proses berulang yang memiliki tingkatan.

### 2.1.6 Fungsi print() dan Parameter end

Fungsi print() digunakan untuk menampilkan informasi atau hasil proses ke layar. Secara default, print() akan berpindah ke baris baru setelah menampilkan suatu nilai.

Pada program digunakan:

print(".", end="")

Parameter end="" digunakan agar setelah mencetak tanda titik, kursor tidak langsung berpindah ke baris berikutnya. Dengan demikian, beberapa tanda titik dapat dicetak secara berurutan pada baris yang sama.

Setelah perulangan kedua selesai, program menjalankan:

print(line)

Perintah tersebut digunakan untuk mencetak nilai line dan kemudian berpindah ke baris berikutnya. Kombinasi kedua penggunaan print() tersebut digunakan untuk membentuk pola keluaran berdasarkan nilai batas yang diberikan.

### 2.2.3 Program perulangan_for3_2611533007

```python
ulang_3007 = int(input("Masukkan jumlah perulangan: "))

jumlah_3007 = 0
for i in range(1, ulang_3007 + 1):
    print(i, end="")
    jumlah_3007 = jumlah_3007 + i

    if i < ulang_3007:
        print(" + ", end="")
    else:
        print(" = ", end="")
print()
print("Jumlah =", jumlah_3007)
```

### 2.2.4 Program nested_for1_2611533007

### 2.2.7 Program nested_for4_2611533007

```python
tinggi_3007 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_3007 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_3007 = tinggi_3007
    c_3007 = a_3007
    lebar_3007 = (2 * tinggi_3007) - 2

    for i in range(1, tinggi_3007 + 1):
        b_3007 = c_3007 + 1

        for j in range(1, lebar_3007 + 1):

            # Baris atas dan bawah
            if i == 1 or i == tinggi_3007:
                if j == 1 or j == lebar_3007:
                    print("#", end="")
                else:
                    print("=", end="")
            # Baris isi
            else:
                if j == 1 or j == lebar_3007:
                    print("|", end="")
                else:
                    if j == c_3007:
                        print("<", end="")
                    elif j == b_3007:
                        print(">", end="")
                    elif j == (lebar_3007 - c_3007):
                        print("<", end="")
                    elif j == (lebar_3007 - c_3007 + 1):
                        print(">", end="")
                    elif j > b_3007 and j < (lebar_3007 - c_3007):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli Java
        a_3007 -= 2

        if a_3007 <= 0:
            c_3007 = (-a_3007) + 2
        else:
            c_3007 = a_3007
```

8.  for j in range(1, lebar_3007 + 1):  melakukan perulangan untuk menentukan setiap posisi karakter dalam satu baris.<br>
