GitHub menyediakan repository sebagai tempat penyimpanan berbagai file proyek, seperti kode program, dokumen, gambar, dan file pendukung lainnya. Setiap proyek dapat memiliki repository sendiri sehingga seluruh file yang berkaitan dengan proyek tersebut dapat disimpan dalam satu tempat.

Salah satu fungsi GitHub adalah membantu menyimpan riwayat perubahan pada suatu proyek. Perubahan yang dilakukan dapat dicatat melalui commit, sehingga pengguna dapat mengetahui perkembangan kode dari waktu ke waktu. GitHub juga memungkinkan pengguna untuk mengunggah atau push file dari komputer ke repository secara daring.

Selain sebagai tempat penyimpanan, GitHub juga dapat digunakan untuk membagikan proyek kepada orang lain. Repository dapat diatur agar dapat diakses oleh pengguna lain sesuai dengan pengaturan yang diberikan. Hal ini membuat GitHub dapat digunakan dalam kegiatan pembelajaran, pengembangan proyek, maupun kerja sama dalam pembuatan perangkat lunak.

Dalam kegiatan praktikum, GitHub dapat digunakan untuk menyimpan hasil program dan laporan. File yang telah dibuat melalui VS Code dapat diunggah ke repository GitHub sehingga hasil pekerjaan memiliki salinan secara daring dan dapat diakses kembali ketika diperlukan.

Penggunaan GitHub juga membantu mahasiswa mengenal pengelolaan proyek pemrograman secara terstruktur. ## 2.2.1 Membuat Repository GitHub
Setelah program selesai dibuat dan dijalankan melalui VS Code, hasil pekerjaan dapat disimpan secara daring dengan memanfaatkan GitHub. GitHub menyediakan repository yang berfungsi sebagai tempat penyimpanan serta pengelolaan file proyek. Untuk mengunggah program dari komputer ke repository GitHub, diperlukan beberapa tahapan sebagai berikut:
1. Membuat sebuah repository baru pada akun GitHub.
2. Membuka folder yang berisi proyek melalui VS Code.
3. Menjalankan terminal yang tersedia pada VS Code.
4. Mengaktifkan Git pada folder proyek dengan melakukan inisialisasi repository lokal.
5. Memasukkan file proyek yang akan disimpan ke dalam staging area.
6. Menyimpan perubahan tersebut dengan membuat sebuah commit.
7. Menetapkan repository GitHub sebagai remote repository dari proyek lokal.
8. Mengirimkan file dan perubahan dari repository lokal ke GitHub melalui proses push.

Contoh perintah yang digunakan :

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

**Penjelasan :**
1. git init digunakan untuk membuat repository Git pada folder proyek.
2. git add . digunakan untuk menambahkan seluruh perubahan ke staging area.
3. git commit digunakan untuk menyimpan perubahan dengan sebuah pesan.
4. git branch -M main digunakan untuk menentukan branch utama dengan nama main.
5. git remote add origin digunakan untuk menghubungkan repository lokal dengan repository GitHub.
6. git push digunakan untuk mengunggah kode dari komputer ke GitHub.
