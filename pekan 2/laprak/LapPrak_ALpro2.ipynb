## 2.1 Dasar Teori
### 2.1.1 Python
Python merupakan salah satu bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana dan mudah dipahami. Python banyak digunakan dalam berbagai bidang, seperti pengembangan perangkat lunak, pengolahan data, dan pembelajaran pemrograman. Karena sintaksnya relatif sederhana, Python juga sesuai digunakan untuk mempelajari konsep dasar pemrograman.

Dalam Python terdapat berbagai sintaks dasar, seperti fungsi print() yang digunakan untuk menampilkan informasi ke layar dan komentar yang digunakan untuk memberikan keterangan pada kode program. Python juga menggunakan string untuk menyimpan atau menampilkan data berupa teks. Beberapa string dapat digabungkan menggunakan operator +. Selain itu, Python memiliki escape character, seperti \\, yang dapat digunakan untuk menghasilkan karakter atau format tertentu dalam output.

### 2.1.2 Variabel 
Variabel merupakan tempat penyimpanan data yang bersifat mutable, artinya nilainya bisa berubah-ubah. Variabel dalam python memiliki format penulisan nama_variabel = . Variabel dapat berisi teks maupun bilangan. Terdapat beberapa aturan penulisan variabel, yaitu :
- nama variabel boleh diawali menggunakan huruf atau garis bawah (_) dan tidak dapat dimulai dengan angka (0-9), contoh: namasaya, _nama
- karakter pada variabel bersifat sensitif, artinya huruf kapital dan huruf kecil memiliki arti yang berbeda. 
- nama variabel tidak boleh menggunakan kata kunci yang sudah ada dalam python

### 2.1.3 Tipe Data

Tipe data dalam Python mendefinisikan jenis nilai yang disimpan dalam variabel dan menentukan operasi yang dapat dilakukan pada data tersebut. Karena Python memperlakukan semuanya sebagai objek, setiap nilai dikaitkan dengan tipe data tertentu.

#### 1. Tipe Data Numerik

Tipe data numerik merupakan tipe data yang digunakan untuk menyimpan nilai berupa angka. Tipe data ini banyak digunakan dalam program yang membutuhkan proses perhitungan matematika.

Dalam Python, tipe data numerik terdiri atas beberapa jenis. Integer digunakan untuk menyimpan bilangan bulat tanpa nilai desimal. Float digunakan untuk menyimpan bilangan yang memiliki nilai desimal. Sementara itu, complex digunakan untuk menyimpan bilangan kompleks yang memiliki bagian bilangan real dan bagian imajiner.

Pada praktikum, tipe data numerik digunakan untuk mengenali perbedaan karakteristik antara bilangan bulat, bilangan desimal, dan bilangan kompleks. Tipe data numerik juga diterapkan dalam program perhitungan, seperti menghitung luas suatu bangun berdasarkan nilai yang diberikan.

#### 2. Tipe Data String

String merupakan tipe data yang digunakan untuk menyimpan data berupa teks atau kumpulan karakter. String dapat digunakan untuk menyimpan berbagai informasi, seperti nama, jenis kelamin, alamat, keterangan, dan informasi lain yang berbentuk teks.

Data string berbeda dengan data numerik karena string digunakan untuk merepresentasikan karakter atau teks, bukan sebagai nilai yang secara langsung digunakan untuk operasi aritmatika. Oleh karena itu, pemilihan tipe data string diperlukan ketika informasi yang akan disimpan berupa teks.

Dalam praktikum, string digunakan untuk menyimpan informasi identitas. Penggunaan string menunjukkan bahwa variabel dalam Python tidak hanya dapat menyimpan angka, tetapi juga dapat menyimpan informasi dalam bentuk teks.

#### 3. Tipe Data Boolean

Boolean merupakan tipe data yang digunakan untuk menyatakan nilai logika. Tipe data boolean hanya memiliki dua kemungkinan nilai, yaitu True dan False. True menunjukkan bahwa suatu kondisi bernilai benar, sedangkan False menunjukkan bahwa suatu kondisi bernilai salah.

Tipe data boolean banyak digunakan dalam proses pengambilan keputusan pada program. Nilai boolean dapat diperoleh dari hasil perbandingan maupun ditentukan secara langsung. Dalam praktikum, boolean diterapkan dalam program yang digunakan untuk memeriksa kondisi kelulusan.

Penggunaan boolean memungkinkan program untuk memberikan keluaran yang berbeda berdasarkan kondisi tertentu. Boolean juga menjadi dasar dalam penggunaan struktur percabangan dan operator logika.

#### 4. Input

Input merupakan proses memasukkan data ke dalam program yang dilakukan oleh pengguna. Dalam Python, input dapat dilakukan menggunakan fungsi input(). Melalui input, program dapat menerima data secara langsung sehingga nilai yang digunakan tidak harus ditentukan sebelumnya oleh programmer.

Data yang dimasukkan oleh pengguna melalui input pada dasarnya diterima sebagai data string. Apabila data tersebut akan digunakan dalam proses perhitungan, maka diperlukan penyesuaian tipe data menjadi tipe numerik, seperti integer atau float.

Pada praktikum, input digunakan untuk menerima nilai yang diperlukan dalam proses perhitungan, seperti panjang, lebar, dan jari-jari lingkaran. Setelah menerima input, program mengolah data tersebut menggunakan operasi aritmatika dan menghasilkan nilai keluaran.

#### 5. Konstanta

Konstanta merupakan nilai yang digunakan dalam program dan dianggap tetap selama program dijalankan. Konstanta biasanya digunakan untuk menyimpan nilai yang tidak perlu mengalami perubahan, seperti nilai matematika tertentu.

Penggunaan konstanta dapat membuat program lebih terstruktur karena nilai yang bersifat tetap dapat dibedakan dari nilai yang dapat berubah. Salah satu contoh konstanta yang digunakan dalam praktikum adalah nilai π (pi) untuk melakukan perhitungan luas lingkaran.

Dalam Python, konsep konstanta dapat ditandai menggunakan Final. Meskipun Python tidak secara khusus melarang perubahan nilai tersebut, penggunaan Final memberikan penanda bahwa nilai tersebut dimaksudkan untuk tetap dan tidak diubah selama program digunakan.

#### 6. Fungsi print() dan type()

Fungsi print() merupakan fungsi yang digunakan untuk menampilkan data atau hasil pengolahan program kepada pengguna. Fungsi ini dapat menampilkan teks, nilai variabel, maupun hasil perhitungan. Dalam praktikum, fungsi print() digunakan untuk menampilkan nilai variabel dan hasil dari proses pengolahan data.

Sementara itu, fungsi type() digunakan untuk mengetahui tipe data yang dimiliki oleh suatu nilai atau variabel. Fungsi ini dapat membantu programmer dalam memastikan jenis data yang sedang digunakan dalam program.

Penggunaan kedua fungsi tersebut mendukung proses pembelajaran tipe data dan variabel. Fungsi type() digunakan untuk mengidentifikasi tipe data, sedangkan print() digunakan untuk melihat hasil atau informasi yang dihasilkan oleh program.

#### 7. Operator Dasar

Operator merupakan simbol atau tanda yang digunakan untuk melakukan operasi terhadap data. Operator memungkinkan program melakukan berbagai proses, seperti perhitungan, perbandingan, dan pengolahan kondisi.

Operator aritmatika digunakan untuk melakukan operasi matematika, seperti penjumlahan, pengurangan, perkalian, dan pembagian. Dalam praktikum, operator aritmatika digunakan untuk menghitung luas persegi panjang dan luas lingkaran.

Operator perbandingan digunakan untuk membandingkan dua nilai. Hasil dari perbandingan tersebut berupa nilai boolean, yaitu True atau False. Pada program pemeriksaan kelulusan, operator perbandingan digunakan untuk membandingkan nilai yang diperoleh dengan batas nilai kelulusan.

Operator logika digunakan untuk menggabungkan atau memproses beberapa kondisi. Salah satu operator logika yang digunakan adalah and, yang menghasilkan kondisi benar apabila kondisi-kondisi yang diperiksa terpenuhi.

Dengan memahami operator dasar, mahasiswa dapat melakukan pengolahan data dan membuat program yang mampu memberikan hasil berdasarkan kondisi tertentu.

### 2.2.3 Program Boolean_2611533007

```python
is_lulus_3007 = True
is_cumlaude_3007 = True

nilai_3007 = 80
batas_lulus_3007 = 75

status_kelulusan_3007 = nilai_3007 >= batas_lulus_3007 

print("=== Check Kelulusan ===")
print("Nilai", nilai_3007)
print("Apakah Lulus?:", status_kelulusan_3007)
if is_lulus_3007 and is_cumlaude_3007:
    print("Selamat, Anda lulus dengan predikat Cum laude!")
```

**Penjelasan :**

2. is_cumlaude_3007 = True, menyimpan status bahwa variabel tersebut memiliki predikat Cum laude.
3. nilai_3007 = 80, menyimpan nilai yang diperoleh, yaitu 80.
4. batas_lulus_3007 = 75, menentukan nilai minimum kelulusan, yaitu 75.
5. status_kelulusan_3007 = nilai_3007 >= batas_lulus_3007, membandingkan nilai 80 dengan batas lulus 75. Hasilnya True karena 80 ≥ 75.
6. print(), menampilkan informasi nilai dan status kelulusan.
7. if is_lulus_3007 and is_cumlaude_3007:, mengecek apakah kedua kondisi bernilai True menggunakan operator and.
8. print("Selamat..."), jika kedua kondisi terpenuhi, program menampilkan ucapan bahwa pengguna lulus dengan predikat Cum laude.

### 2.2.5 Program Konstanta_2611533007

```python
from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI)) 
jari_3007 = float(input('Masukkan  nilai jari-jari:'))
luas_3007 = PI * jari_3007 * jari_3007
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_3007, luas_3007))
```

**Penjelasan :**
1. from typing import final, mengimpor Final untuk menandai sebuah variabel sebagai konstanta.
2. PI: Final = 3.14, menetapkan nilai konstanta PI sebesar 3,14.
3. print("pi: %f" % (PI)), menampilkan nilai PI dalam format bilangan desimal.
4. jari_3007 = float(input(...)), meminta pengguna memasukkan jari-jari dan mengubah input menjadi tipe float.
5. luas_3007 = PI * jari_3007 * jari_3007, menghitung luas lingkaran menggunakan rumus π × r × r.
6. print(...), menampilkan nilai jari-jari dan hasil luas lingkaran dengan 2 angka di belakang koma.
